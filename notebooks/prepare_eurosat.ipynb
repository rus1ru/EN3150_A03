# Download and prepare EuroSAT RGB

Run the cells in order. The first preparation downloads the official 27,000-image EuroSAT RGB archive from Zenodo, verifies its published MD5 checksum, safely extracts the images to `data/raw/EuroSAT_RGB`, and uses the project's existing preparation code to validate the dataset and prepare the canonical split in `data/splits_torch_v1/`.

The committed split manifests are the source of truth; this notebook does not define a new split. Raw images and the ZIP are git-ignored and should not be committed. Select a Jupyter kernel with this project's Python dependencies installed. The first download requires internet access and sufficient disk space.

```python
from pathlib import Path
import subprocess
import sys

PROJECT_ROOT = next(
    (parent for parent in (Path.cwd(), *Path.cwd().parents)
     if (parent / "scripts" / "prepare_data.py").is_file()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError(
        "Could not locate scripts/prepare_data.py. Start Jupyter from this repository "
        "or one of its subdirectories."
    )

PREPARE_SCRIPT = PROJECT_ROOT / "scripts" / "prepare_data.py"
SPLIT_DIR = PROJECT_ROOT / "data" / "splits_torch_v1"
print(f"Project root: {PROJECT_ROOT}")
print(f"Images: {PROJECT_ROOT / 'data' / 'raw' / 'EuroSAT_RGB'}")
print(f"Canonical split: {SPLIT_DIR}")
```

## Download images and prepare the split

This runs the repository's official preparation script. It checks the archive before extraction and respects the existing canonical split rather than silently replacing it.

```python
subprocess.run(
    [
        sys.executable,
        str(PREPARE_SCRIPT),
        "--download",
        "--split-dir",
        str(SPLIT_DIR),
    ],
    cwd=PROJECT_ROOT,
    check=True,
)
```

## Verify the downloaded images and committed manifests

Run the explicit verification as a final audit. A successful result confirms the local images agree with the project's committed split manifests.

```python
subprocess.run(
    [
        sys.executable,
        str(PREPARE_SCRIPT),
        "--verify-only",
        "--split-dir",
        str(SPLIT_DIR),
    ],
    cwd=PROJECT_ROOT,
    check=True,
)
print("EuroSAT images and canonical split verified successfully.")
```
